# Tuples

A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

Characterstics

- Ordered
- Unchangeble
- Allows duplicate

### Plan of attack

- Creating a Tuple
- Accessing items 
- Editing items
- Adding items
- Deleting items
- Operations on Tuples
- Tuple Functions

### Creating Tuples

```python
# empty
t1=()
print(t1)
# create a tuple with a single item
t2 = ('hello',)
print(t2)
print(type(t2))
#homo
t3 = (1,2,3,4,)
print(t3)
#hetro
t4 = (1,2,3,True,[1,2,3])
print(t4)
#tuple
t5 = (1,2,3,(4,6))
print(t5)
# usong type conversion 
t6=tuple('hello')
print(t6)
```

## operations on tuple

```python
# + and -
t1=(1,2,3,4)
t2=(5,6,7,8)
print(t1+t2)
print(t1*5)
# memebership 
1 in t1

# iteration 
for i in t1:
    print(i)

```

## tuple functions

```python
# len/sum/min/max/sorted
t = (1,2,3,4)
print(len(t))
print(max(t))
print(sum(t))

print(min(t))
print(max(t))
print(sorted(t,reverse=True))
```

```python
#count
t =(1,2,5,4,3,5)

print(t.count(5))
print(t.count(50)) 
```

```python
# index
t.index(5)
#t.index(50)  #ValueError: tuple.index(x): x not in tuple
```

### difference between list and tuples 
- Syntax
- Mutability
- speed
- memory= list more
- builtin functionality = list has more 
- error prone = list more
- usability 


```python
import time 
L = list(range(100000000))
T = tuple(range(100000000))
start = time.time()
for i in L:
    i*5
print('list time',time.time()-start)
start = time.time()
for i in T:
    i*5
print('tuple time',time.time()-start)
```

```python
import sys 
L = list(range(1000))
T = tuple(range(1000))
print('List Size',sys.getsizeof(L))
print('Tuple Size',sys.getsizeof(T))
```

## special syntax 

```python
# tuple unpacking
a,b,c =(1,2,3)
print(a,b,c)
```

```python
a,b =(1,2,3)
print(a,b,c)
```

```python
a=1
b=2
a,b=b,a
print(a,b)
```

```python
a,b,*others =(1,2,3,6,8)
print(a,b)
print(others)
```

```python
#zipping tuples
a=(1,2,3,4)
b = (6,89,9,5)
print(zip(a,b))
print(tuple(zip(a,b)))


```
