# Tuples

A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

Characterstics

- Ordered
- Unchangeble
- Allows duplicate

# Sets

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

Characterstics:
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

### Creating Sets

```python
# empty set
s={} # not a set its a dictionary because set is mutable and dictionary is also mutable so python will consider it as a dictionary
s = set()
print(s)
print(type(s))
#1D and 2D set
s1={1,2,3,4}
print(s1)
s2 = {1,2,3,4,(5,6)}
print(s2)
 #s3={1,2,3,4,{5,6}}# not create 2d set because set can't have mutable elements like list, dict, set
# print(s3)
# homogeneous and heterogeneous set
s4 ={1,'hello',True,5.6,(5,9),1}
print(s4)
# using type conversion
s5 = set('hello')
print(s5)
s8 = set([1,2,3,4,5])
print(s8)
#duplicates are not allowed in set
s6 = {1,2,3,4,5,6,7,8,9,1,2,3}
print(s6)
#set can't have mutable elements like list, dict, set
  #s7 = {1,2,3,4,[5,6]} #TypeError: unhashable type: 'list'
# print(s7)
```

```python
s1 ={1,2,3,4,5}
s2= {5,4,3,2,1}
print(s1==s2)  # because set is unordered collection of unique elements so order doesn't matter in set
```

## set operation

```python
# union - (|)
s1 = {1,2,3,4}
s2 = {3,4,5,6}  
print(s1|s2) # {1, 2, 3, 4, 5, 6}
#intersection - (&)
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1&s2) # {3, 4}   
# difference - (-)
s1 = {1,2,3,4}  
s2 = {3,4,5,6}
print(s1-s2) # {1, 2}
print(s2-s1) # {5, 6}
#symmetric difference - (^)
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1^s2) # {1, 2, 5, 6} 
# membership - (in)
s1 = {1,2,3,4}  
print(2 in s1) # True
print(5 in s1) # False
#iteration
s1 = {1,2,3,4}
for i in s1:
    print(i)    
```

```python
# len/sum/min/max/sorted
s = {1,2,3,4}
print(len(s))
print(sum(s))
print(min(s))   
print(max(s))
print(sorted(s,reverse=True))
print(sorted(s,reverse=False))
print(sorted(s))
```

```python
# union/update
s1 = {1,2,3,4}
s2 = {3,4,5,6}
s1.union(s2) # {1, 2, 3, 4, 5, 6}
s1.update(s2) 
print(s1) # {1, 2, 3, 4, 5, 6}
print(s2)
```

```python
# intersection/intersection_update
s1 = {1,2,3,4}
s2 = {3,4,5,6}

print(s1.intersection(s2)) # {3, 4}

s1.intersection_update(s2) # {3, 4}
print(s1) # {3, 4}
print(s2) # {3, 4, 5, 6}
```

```python
# difference/difference_update
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1.difference(s2)) # {1, 2}
s1.difference_update(s2) # {1, 2}
print(s1) # {1, 2}
print(s2) # {3, 4, 5, 6}
```

```python
# symmetric_difference/symmetric_difference_update
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1.symmetric_difference(s2)) # {1, 2, 5, 6}
s1.symmetric_difference_update(s2) # {1, 2, 5, 6}
print(s1) # {1, 2, 5, 6}
print(s2) # {3, 4, 5, 6}
```

```python
# isdisjoint/issubset/issuperset
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1.isdisjoint(s2)) # False
print(s1.issubset(s2)) # False
print(s1.issuperset(s2)) # False
# other ex 
s1 = {1,2,3,4}
s2 = {3,4,}
print(s1.isdisjoint(s2)) # False
print(s1.issubset(s2)) # False
print(s1.issuperset(s2)) # True
```

```python
# copy
s1 = {1,2,3,4}
s2 = s1.copy() # address alag hota dono ka 
print(s1)
print(s2)
```

### frozen set 
- frozen set is just an immutable version of a python set object 

```python
# create  frozen set
fs = frozenset([1,2,3,4])
print(fs)
```

```python
# what works and what doesn't work in frozen set
# work all read functions
# doesn't work all write functions
# union ,intersection, difference, symmetric_difference, isdisjoint, issubset, issuperset, copy works in frozen set
# methods that don't work in frozen set: add, remove, discard, pop, clear, update, intersection_update, difference_update, symmetric_difference_update
fs1 = frozenset([1,2,3,4])
fs2 = frozenset([3,4,5,6])
print(fs1.union(fs2)) # {1, 2, 3, 4, 5, 6}
print(fs1.intersection(fs2)) # {3, 4}
print(fs1.difference(fs2)) # {1, 2}
print(fs1.symmetric_difference(fs2)) # {1, 2, 5, 6}
```

# Dictionary

Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as map or assosiative arrays.

dict = { 'name' : 'nitish' , 'age' : 33 , 'gender' : 'male' }

Characterstics:

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

# create dictionary

```python
# empty dictionary 
d={}
print(d)
# 1d dictionary
d1 = {'a':1,'b':2,'c':3}
print(d1)
# with mixed keys
d2 = {1:'a',2:'b','name':'annu','age':20,(1,2,3):1}
print(d2)
# 2d dictionary -> json format
s ={
    'name':'annu',
    'college':'abc',
    'sem':5,
    'subjects':{
        'sub1':'python',
        'sub2':'java',
        'sub3':'c++'
    }
}
print(s)
# using sequence and dict function
d4 = dict([(1,'a'),(2,'b'),(3,'c')])
print(d4)
#duplicate keys 
d5 = {'a':1,'a':2,'a':3}
print(d5)
# mutable items as keys
d7 = {1:'a',(4,5,6):'b'}
print(d7)
d6 = {[1,2,3]:'a',(4,5,6):'b'}
print(d6)

```

## Accessing items

```python
dict1 = {'a':1,'b':2,'c':3}
#[]
print(dict1['a']) # 1
#get
dict1.get('a') # 1
dict1.get('d') # None
```

# adding key-value pair 

```python
dict1 = {'a':1,'b':2,'c':3}
dict1['gender']= 'female'
print(dict1)

# in 2d dictionary
dict2 = {'person1': {'name': 'annu', 'age': 20}, 'person2': {'name': 'john', 'age': 25}}    
print(dict2)
# add a new key-value pair to the inner dictionary
dict2['person1']['gender'] = 'female'
print(dict2)
```

## remove key-value pair

```python
d={'name':'annu','age':20,'gender':'female'}
#pop 
d.pop('age')
print(d)
#popitem - remove and return an arbitrary item from the dictionary. Raises KeyError if the dict is empty.
d.popitem()
print(d)
#del - remove a key-value pair from the dictionary. Raises KeyError if the key is not found.
d = {'name':'annu','age':20,'gender':'female'}
del d['age']
print(d)
#clear - remove all items from the dictionary.
d.clear()
print(d)
del d 
print(d) # NameError: name 'd' is not defined
```

```python
# delete in 2d dictionary
dict2 = {'person1': {'name': 'annu', 'age': 20}, 'person2': {'name': 'john', 'age': 25}}
del dict2['person1']['age'] 
print(dict2)
```

```python
# editing key value pair in 2d dictionary
dict2 = {'person1': {'name': 'annu', 'age': 20}, 'person2': {'name': 'john', 'age': 25}}
dict2['person1']['age'] = 21    
print(dict2)
```

### dictionary operation

```python
# membership - check if a key is present in the dictionary. Returns True if the key is present, False otherwise.
d = {'name':'annu','age':20,'gender':'female'}
print('name' in d) # True
print('height' in d) # False
```

```python
#iteration - iterate over the keys of the dictionary using a for loop.
d = {'name':'annu','age':20,'gender':'female'}
for key in d:
    print(key, d[key])  
```

### dictionary function

```python
#len
print(d)
print(len(d)) # 3

# sorted- sort the keys of the dictionary
print(sorted(d)) # ['age', 'gender', 'name']

#max 
print(max(d)) # name
#min
print(min(d)) # age
```

```python
# items() - returns a view object that displays a list of a dictionary's key-value tuple pairs.
d = {'name':'annu','age':20,'gender':'female'}
for key, value in d.items():
    print(key, value)
```

```python
#keys() - returns a view object that displays a list of all the keys in the dictionary.
d = {'name':'annu','age':20,'gender':'female'}
print(list(d.keys())) # ['name', 'age', 'gender']
print(d.items()) # dict_items([('name', 'annu'), ('age       
```

```python
#values() - returns a view object that displays a list of all the values in the dictionary.
d = {'name':'annu','age':20,'gender':'female'}
print(list(d.values())) # ['annu', 20, 'female']
print(d.values())   
```

```python
# update() - update the dictionary with the key-value pairs from another dictionary or from an iterable of key-value pairs.
d1 = {'name':'annu','age':20}
d2 = {'gender':'female'}
d1.update(d2)
print(d1) # {'name': 'annu', 'age': 20, 'gender': 'female'} 
```

### dict comprehension

```python
# {key:value for vars in iterable} - dictionary comprehension
# example
#print 1st 10 no and their squares in dictionary format
{num: num**2 for num in range(1, 11)}
```

```python
# using existing dictionary to create a new dictionary with modified values
d = {'a': 1, 'b': 2, 'c': 3}
d_new = {key: value**2 for key, value in d.items()} 
print(d_new) # {'a': 1, 'b': 4, 'c': 9}
```

```python
# using zip
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
temp = [30, 32, 31, 29, 28]
{i: j for (i, j) in zip(days, temp)}
```

```python
# using if condition in dictionary comprehension
{i: j for (i, j) in zip(days, temp) if j > 30}
```

```python
# nested dictionary comprehension
#print tables of 1 to 10 in dictionary format
{num: {i: num*i for i in range(1, 11)} for num in range(1, 11)}
```
