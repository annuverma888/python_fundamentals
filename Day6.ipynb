### types of arguments 
- Default argument
- Positional Argument 
- Keyword Argument 

```python
## def argument 
def power(a=1,b=1):
    print(a**b)
power()
power(2,3)
```

```python
#postional argument - the values passed to the function are assigned to the parameters in the order they are passed.
power(2,3) # 8 mila 9 ku ni mila 
# position follow krta pahla pahle ko milta dusara dusare ko milta order matter
```

```python
# keyword argument - the values passed to the function are assigned to the parameters by their name, order does not matter.
power(a=2,b=3)
power(b=3,a=2) 
```

## *args and **kwargs
- are special python keywords that are used to pass the variable length of argument to a function

```python
# *args - it allow us to pass the variable no of non-keyword arguments to a function. it is used when we are not sure how many arguments(input values) will be passed to a function.
def mul(a,b):
    return a*b
mul(2,3)
```

```python
def mul(a,b,c):
    return a*b*c
mul(2,3,4)
```

```python
def mul(*args):# args ke placepr koi bhi word use kr sakte hai
    product = 1
    for i in args:
        product = product*i
    return product
mul(1,2,3,4,45,)
mul(3,4,5,7)
```

```python
#  **kwargs - allows us to pass any no of keyword arguments 
# keyword arguments mean that contain  a key value pair like a python dictionary 
```

```python
def display(**kwargs): # kwargs ke place pr kuch bhi likh sakte
    for (key,value) in kwargs.items():
        print(key,"->",value)
```

```python
display(india="delhi",srilanka="colambo", pakistan="islamabad",nepal="kathmandu")
```

### points to remember while using *args and **kwargs
- order of the arguments matter (normsl -> *args -> **kwargs)
- the words "args" and "kwargs" are only a convention , you can use any name of your choice
